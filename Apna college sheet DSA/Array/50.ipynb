### Brute Force

```python
class Solution(object):
   def myPow(self, x, n):
       """
       :type x: float
       :type n: int
       :rtype: float
       """
       ans = 1.0  # Initialize result to 1.0
       power = abs(n)  # Take the absolute value of n to handle negative exponents


       # Multiply x by itself 'power' times (equivalent to multiplying x^|n| times)
       for _ in range(power):  # Loop runs O(N) times
           ans *= x  # Multiply the current value of ans by x
      
       # If n is negative, return the reciprocal (1/ans). Otherwise, return the result as is.
       return 1.0 / ans if n < 0 else ans




# Main function to test the myPow function
if __name__ == "__main__":
   solution = Solution()  # Create an instance of Solution
   x = 2.0  # Example base
   n = 3  # Example exponent
   result = solution.myPow(x, n)  # Call the myPow method
   print(f"{x}^{n} = {result}")  # Print the result of x raised to the power of n


```

### Better

```python
class Solution(object):
   # Function to calculate x raised to the power of n
   def myPow(self, x, n):
       """
       :type x: float
       :type n: int
       :rtype: float
       """
       return pow(x, n)  # Use the built-in pow function


# Main function to test the myPow function
if __name__ == "__main__":
   solution = Solution()  # Create an instance of Solution
   x = 2.0  # Base value
   n = 3  # Exponent value
   print("Result:", solution.myPow(x, n))  # Print the result
```

### Best

```python

class Solution(object):
   # Helper method for exponentiation by squaring
   def helper(self, x, n):
       if n == 0:
           return 1  # Base case: x^0 = 1
       temp = self.helper(x, n // 2)  # Recursively calculate x^(n//2)
       temp *= temp  # Square the result
       if n % 2 != 0:
           temp *= x  # If exponent is odd, multiply once more by x
       return temp


   def myPow(self, x, n):
       if n < 0:
           val = -n  # Convert negative exponent to positive
           ans = self.helper(x, val)
           return 1.0 / ans  # Return reciprocal for negative exponent
       return self.helper(x, n)


# Dummy input
x = 2.0  # Base
n = 10   # Exponent


sol = Solution()
print(sol.myPow(x, n))  # Output: 1024.0
```
