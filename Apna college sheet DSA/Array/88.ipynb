```python
class Solution(object):
   def merge(self, nums1, m, nums2, n):
       """
       :type nums1: List[int]
       :type m: int
       :type nums2: List[int]
       :type n: int
       :rtype: None. Do not return anything, modify nums1 in-place instead.
       """
       # Copy elements from nums2 to nums1 starting from index m
       for i in range(m, m + n):
           nums1[i] = nums2[i - m]


       # Sort the merged array
       nums1.sort()




# Run the test
if __name__ == "__main__":
   # Example input
   nums1 = [1, 2, 3, 0, 0, 0]  # nums1 has extra space
   m = 3  # Actual elements in nums1
   nums2 = [2, 5, 6]
   n = 3  # Elements in nums2


   # Create object and call the method
   sol = Solution()
   sol.merge(nums1, m, nums2, n)


   # Print the merged and sorted array
   print("Merged array:", nums1)


```
