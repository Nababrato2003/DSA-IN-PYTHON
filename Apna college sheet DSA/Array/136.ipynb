### Naive

```python
class Solution(object):
   def singleNumber(self, nums):
       """
       :type nums: List[int]
       :rtype: int
       """
       n = len(nums)


       # For each element, count how many times it appears
       for i in range(n):
           count = 0
           for j in range(n):
               if nums[i] == nums[j]:
                   count += 1


           # If it appears exactly once, return it
           if count == 1:
               return nums[i]


       # If no such element found
       return -1  # Should never reach here




# Run the test case
if __name__ == "__main__":
   # Test case: all elements appear twice except 4
   nums = [2, 3, 5, 4, 5, 3, 2]


   sol = Solution()
   result = sol.singleNumber(nums)


   print("Single number is:", result)
```

### Better


```python
class Solution(object):
   def singleNumber(self, nums):
       """
       :type nums: List[int]
       :rtype: int
       """
       freq = {}  # Dictionary to store frequency


       # Count occurrences of each number
       for num in nums:
           freq[num] = freq.get(num, 0) + 1


       # Find the number that appears only once
       for key, value in freq.items():
           if value == 1:
               return key


       return -1  # Should never reach here




# Run the test
if __name__ == "__main__":
   # Test case: all elements appear twice except 4
   nums = [2, 3, 5, 4, 5, 3, 2]


   sol = Solution()
   result = sol.singleNumber(nums)


   print("Single number is:", result)
```

### Best


```python
class Solution(object):
   def singleNumber(self, nums):
       """
       :type nums: List[int]
       :rtype: int
       """
       xorResult = 0


       # XOR all elements: duplicates cancel out, single remains
       for num in nums:
           xorResult ^= num  # XOR operation


       return xorResult




# Run the test
if __name__ == "__main__":
   # Test case: all elements appear twice except 4
   nums = [2, 3, 5, 4, 5, 3, 2]


   sol = Solution()
   result = sol.singleNumber(nums)


   print("Single number is:", result)
```
