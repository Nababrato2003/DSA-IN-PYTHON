## Brute force 


```python
class Solution(object):
   def maxSubArray(self, nums):
       """
       :type nums: List[int]
       :rtype: int
       """
       n = len(nums)  # Get the number of elements in the input list
       max_sum = float('-inf')  # Initialize maximum sum to negative infinity


       # Outer loop selects the starting index of the subarray
       for i in range(n):
           curr_sum = 0  # Initialize current subarray sum


           # Inner loop extends the subarray and calculates the sum
           for j in range(i, n):
               curr_sum += nums[j]  # Add current element to the current sum
               max_sum = max(max_sum, curr_sum)  # Update maximum sum if current sum is greater


       return max_sum  # Return the maximum subarray sum found




# Main function to test the maxSubArray method
if __name__ == "__main__":
   nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]  # Example input array
  
   solution = Solution()  # Create an instance of the Solution class
   result = solution.maxSubArray(nums)  # Call the maxSubArray method
  
   print("The maximum subarray sum is:", result)  # Output the result



```

### Better approach

```python
class Solution(object):
   def maxSubArray(self, nums):
       """
       :type nums: List[int]
       :rtype: int
       """
       n = len(nums)  # Get the number of elements in the input list
       dp = [0] * n  # dp[i] stores the maximum subarray sum ending at index i
       dp[0] = nums[0]  # Base case: the first element is the only subarray at index 0
       max_sum = nums[0]  # Initialize the maximum sum with the first element


       # Iterate through the list starting from the second element
       for i in range(1, n):
           # Decide whether to start fresh at nums[i] or extend the previous subarray
           dp[i] = max(nums[i], nums[i] + dp[i - 1])


           # Update the maximum sum found so far
           max_sum = max(max_sum, dp[i])


       return max_sum  # Return the maximum subarray sum found




# Main block to test the function
if __name__ == "__main__":
   nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]  # Example input list
   solution = Solution()
   result = solution.maxSubArray(nums)
   print("The maximum subarray sum is:", result)  # Expected output: 6

```

## Best Approach (Kadane's Algo)

```python
class Solution(object):
    def maxSubArray(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """

        # Initialize current sum and maximum sum
        currSum = 0
        maxSum = float('-inf')   # Equivalent to INT_MIN

        # Traverse the array
        for i in range(len(nums)):

            # Add current element to the current sum
            currSum += nums[i]

            # Update maximum sum if current sum is greater
            maxSum = max(maxSum, currSum)

            # If current sum becomes negative, reset it to 0
            if currSum < 0:
                currSum = 0

        return maxSum


# Example usage
if __name__ == "__main__":

    nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

    solution = Solution()

    result = solution.maxSubArray(nums)

    print("Maximum Subarray Sum:", result)
```
