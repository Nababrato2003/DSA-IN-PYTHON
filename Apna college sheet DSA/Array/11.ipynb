### Brute force apporach 

```python
class Solution(object):
    def maxArea(self, height):
        """
        :type height: List[int]
        :rtype: int
        """

        # Initialize maximum water (maximum area)
        maxWater = 0

        n = len(height)

        # Choose the left boundary
        for i in range(n):

            # Choose the right boundary
            for j in range(i + 1, n):

                # Width between the two lines
                width = j - i

                # Height is the smaller of the two lines
                ht = min(height[i], height[j])

                # Area formed
                area = width * ht

                # Update maximum area
                maxWater = max(maxWater, area)

        return maxWater


# Driver Code
if __name__ == "__main__":

    height = [1, 8, 6, 2, 5, 4, 8, 3, 7]

    solution = Solution()

    print("Maximum Water =", solution.maxArea(height))
```

### Optimised approach(2pointer approach)

```python
class Solution(object):
    def maxArea(self, height):
        """
        :type height: List[int]
        :rtype: int
        """

        n = len(height)

        # Initialize pointers and answer
        left_pointer = 0
        right_pointer = n - 1
        max_water = 0

        # Continue until both pointers meet
        while left_pointer < right_pointer:

            # Calculate width
            width = right_pointer - left_pointer

            # Calculate height
            ht = min(height[left_pointer], height[right_pointer])

            # Calculate current area
            current_water = width * ht

            # Update maximum water
            max_water = max(max_water, current_water)

            # Move the pointer having smaller height
            if height[left_pointer] < height[right_pointer]:
                left_pointer += 1
            else:
                right_pointer -= 1

        return max_water


# Driver Code
if __name__ == "__main__":

    height = [1, 8, 6, 2, 5, 4, 8, 3, 7]

    solution = Solution()

    print(solution.maxArea(height))
```
